```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

data = df[["Quantity","Sales"]].dropna()
X = data[["Quantity"]]
y = data["Sales"]

model = LinearRegression()
model.fit(X, y)
y_pred = model.predict(X)

print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])
print("R2:", model.score(X,y))
print("MSE:", mean_squared_error(y,y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y,y_pred)))

quantity_value = 4
print("Predicted Sales for Quantity 4:",
      model.predict([[quantity_value]])[0])

plt.scatter(X["Quantity"], y)
plt.plot(X["Quantity"], y_pred)
plt.title("Quantity vs Sales Regression")
plt.xlabel("Quantity"); plt.ylabel("Sales")
plt.show()
```

```python
population = df["Sales"].dropna().values
np.random.seed(10)

for size in [30,50,100]:
    sample_means = []
    for i in range(100):
        sample = np.random.choice(population, size=size, replace=False)
        sample_means.append(sample.mean())
    sample_means = np.array(sample_means)

    print("Sample Size:", size)
    print("Mean of Sample Means:", sample_means.mean())
    print("SD:", sample_means.std(ddof=1))
    print("Standard Error:",
          population.std(ddof=1)/np.sqrt(size))

    plt.hist(sample_means, bins=15)
    plt.title("Sampling Distribution - Sample Size " + str(size))
    plt.xlabel("Sample Mean"); plt.ylabel("Frequency")
    plt.show()
```

```python
from scipy import stats
sales = df["Sales"].dropna()
n = len(sales)
mean = sales.mean()
std = sales.std()

for confidence in [0.90,0.95,0.99]:
    alpha = 1-confidence
    t_value = stats.t.ppf(1-alpha/2, n-1)
    margin = t_value * std/np.sqrt(n)
    print(confidence, mean-margin, mean+margin)
```

```python
sales = df["Sales"].dropna().values
np.random.seed(10)
bootstrap_medians = []

for i in range(1000):
    sample = np.random.choice(sales, size=len(sales), replace=True)
    bootstrap_medians.append(np.median(sample))

bootstrap_medians = np.array(bootstrap_medians)
median = np.median(sales)
lower = np.percentile(bootstrap_medians, 2.5)
upper = np.percentile(bootstrap_medians, 97.5)

print("Original Median:", median)
print("Bootstrap Samples:", 1000)
print("95% CI:", lower, upper)

plt.hist(bootstrap_medians, bins=30)
plt.axvline(lower, linestyle="--")
plt.axvline(upper, linestyle="--")
plt.axvline(median)
plt.title("Bootstrap Distribution of Median Sales")
plt.show()
```

```python
from scipy.stats import ttest_1samp
sales = df["Sales"].dropna()
benchmark = 150

t_stat, p_value = ttest_1samp(sales, benchmark)
print("Test Statistic:", t_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Decision: Reject H0")
else:
    print("Decision: Do not reject H0")
```

```python
from scipy.stats import ttest_ind

A = df[df["Discount"] == 0]["Sales"]
B = df[df["Discount"] > 0]["Sales"]

print("Group A Orders:", len(A))
print("Group B Orders:", len(B))
print("Group A Mean Sales:", A.mean())
print("Group B Mean Sales:", B.mean())

t, p = ttest_ind(A, B, equal_var=False)
print("t-statistic:", t)
print("p-value:", p)

plt.bar(["A: No Discount","B: Discount"], [A.mean(),B.mean()])
plt.ylabel("Mean Sales")
plt.title("A/B-Style Sales Comparison")
plt.show()
```

```python
np.random.seed(10)
demand = np.random.choice(df["Quantity"].values, size=30, replace=True)

inventory_levels = [2,3,4]
results = []

for inventory in inventory_levels:
    total_sales = np.minimum(inventory,demand).sum()
    shortage = np.maximum(demand-inventory,0).sum()
    leftover = np.maximum(inventory-demand,0).sum()
    service = total_sales/(total_sales+shortage)*100

    results.append([inventory,total_sales,shortage,leftover,service])

results_df = pd.DataFrame(
    results,
    columns=["Inventory Level","Total Sales",
             "Total Shortage","Total Leftover","Service Level (%)"]
)
print(results_df)

results_df.plot(x="Inventory Level",
                y=["Total Sales","Total Shortage","Total Leftover"],
                kind="bar")
plt.title("Campus Cafe Inventory Scenario Comparison")
plt.ylabel("Units")
plt.show()
```

```python
from scipy.stats import f_oneway

groups = [
    group["Sales"].values
    for _, group in df.groupby("Category")
]

f_statistic, p_value = f_oneway(*groups)

print("F-statistic:", f_statistic)
print("p-value:", p_value)

if p_value < 0.05:
    print("Decision: Reject H0")
else:
    print("Decision: Do not reject H0")
```

```python
import numpy as np
import statsmodels.api as sm
from statsmodels.formula.api import ols

df["QtyLevel"] = np.where(df["Quantity"] <= 3,
                          "LowQty", "HighQty")
df["DiscLevel"] = np.where(df["Discount"] <= 0.05,
                           "LowDiscount", "HighDiscount")

print(df.groupby(["QtyLevel","DiscLevel"]).Sales.mean())

model = ols(
    "Sales ~ C(QtyLevel)*C(DiscLevel)",
    data=df
).fit()

print(sm.stats.anova_lm(model, typ=2))
```
